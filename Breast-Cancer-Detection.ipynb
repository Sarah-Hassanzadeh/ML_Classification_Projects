# preprocessing


```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(bc.data,bc.target,test_size=0.2)
```

```python
print(f'feature -> train: {X_train.shape} | test: {X_test.shape}')
print(f'label -> train: {y_train.shape} | test: {y_test.shape}')
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,1))

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train[0]
```

# classification


```python
from sklearn.metrics import accuracy_score,precision_score,recall_score

def calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test):
    acc_train = accuracy_score(y_true=y_train,y_pred=y_pred_train)
    acc_test = accuracy_score(y_true=y_test,y_pred=y_pred_test)
    
    p = precision_score(y_true=y_test,y_pred=y_pred_test)
    r = recall_score(y_true=y_test,y_pred=y_pred_test)
    
    print(f'acc train : {acc_train} | acc test : {acc_test} | precision : {p} | recall : {r}')
    
    return acc_train,acc_test,p,r
```

# GaussianNB

```python
from sklearn.naive_bayes import GaussianNB

gnb = GaussianNB()
gnb.fit(X_train,y_train)
```

```python
y_pred_train = gnb.predict(X_train)
y_pred_test = gnb.predict(X_test)

acc_train_gnb,acc_test_gnb,p_gnb,r_gnb = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# knn

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=8,algorithm='kd_tree',leaf_size=28)
knn.fit(X_train,y_train)
```

```python
y_pred_train = knn.predict(X_train)
y_pred_test = knn.predict(X_test)

acc_train_knn,acc_test_knn,p_knn,r_knn = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# DecisionTree

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(max_depth=64,min_samples_split=2,criterion='gini')
dt.fit(X_train,y_train)
```

```python
y_pred_train = dt.predict(X_train)
y_pred_test = dt.predict(X_test)

acc_train_dt,acc_test_dt,p_dt,r_dt = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# RandomForest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=1000,max_depth=32,min_samples_split=4)
rf.fit(X_train,y_train)
```

```python
y_pred_train = rf.predict(X_train)
y_pred_test = rf.predict(X_test)

acc_train_rf,acc_test_rf,p_rf,r_rf = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# Svm

```python
from sklearn.svm import SVC

svm = SVC(kernel='poly')
svm.fit(X_train,y_train)
```

```python
y_pred_train = svm.predict(X_train)
y_pred_test = svm.predict(X_test)

acc_train_svm,acc_test_svm,p_svm,r_svm = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# LogisticRegression

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(X_train,y_train)
```

```python
y_pred_train = lr.predict(X_train)
y_pred_test = lr.predict(X_test)

acc_train_lr,acc_test_lr,p_lr,r_lr = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```

# Ann

```python
from sklearn.neural_network import MLPClassifier

ann = MLPClassifier(hidden_layer_sizes=256,activation='relu',solver='adam',batch_size=64)
ann.fit(X_train,y_train)
```

```python
y_pred_train = ann.predict(X_train)
y_pred_test = ann.predict(X_test)

acc_train_ann,acc_test_ann,p_ann,r_ann = calcucate_metrics(y_train,y_test,y_pred_train,y_pred_test)
```
